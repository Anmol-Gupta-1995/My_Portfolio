```python
"""
House Price Dataset — Raw CSV -> Steps 1-7 -> PCA + Chi2 -> Model Comparison
==============================================================================
Steps 1-7 are IDENTICAL to the full raw-data pipeline (cleaning, feature
engineering, target log-transform, split, imputation, skewed-feature
log-transform). From there, instead of direct one-hot encoding, this version
applies PCA to the numeric block and Chi2 (SelectKBest) to the categorical
block, searches for the optimal n_components / k, then trains RandomForest
and XGBoost (both default AND tuned) on the reduced feature set.

EDIT THE TWO LINES MARKED "<-- EDIT" BELOW: the CSV path and the target column
name, if it differs from "SalePrice".
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, KFold, cross_val_score, RandomizedSearchCV
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

try:
    from xgboost import XGBRegressor
    HAS_XGB = True
except ImportError:
    HAS_XGB = False
    print("xgboost not installed — will only run RandomForest. `pip install xgboost` to enable it.")

RANDOM_STATE = 42

# ==================================================================
# STEPS 1-7 — identical to the raw-data pipeline
# ==================================================================

# ------------------------------------------------------------------
# 1. Load the RAW dataset
# ------------------------------------------------------------------
CSV_PATH = '/Users/anmolgupta/Desktop/Datamites Internship/House_price_prediction (PRCP-1020)/data.csv'         # <-- EDIT: path to your raw file
TARGET = "SalePrice"                        # <-- EDIT: if your target column is named differently

df = pd.read_csv(CSV_PATH)
print(f"Raw data shape: {df.shape}")
print("Missing values per column (top 15):")
print(df.isnull().sum().sort_values(ascending=False).head(15))

# ------------------------------------------------------------------
# 2. Missing-value handling
# ------------------------------------------------------------------
NONE_MEANS_ABSENT_CAT = [
    "Alley", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2",
    "FireplaceQu", "GarageType", "GarageFinish", "GarageQual", "GarageCond",
    "PoolQC", "Fence", "MiscFeature", "MasVnrType",
]
ZERO_MEANS_ABSENT_NUM = [
    "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2", "BsmtUnfSF", "TotalBsmtSF",
    "BsmtFullBath", "BsmtHalfBath", "GarageYrBlt", "GarageCars", "GarageArea",
]
for col in NONE_MEANS_ABSENT_CAT:
    if col in df.columns:
        df[col] = df[col].fillna("None")
for col in ZERO_MEANS_ABSENT_NUM:
    if col in df.columns:
        df[col] = df[col].fillna(0)

print(f"\nRemaining missing values after explicit None/0 fill: {df.isnull().sum().sum()}")

# ------------------------------------------------------------------
# 3. Feature engineering
# ------------------------------------------------------------------
def safe_cols_exist(cols):
    return all(c in df.columns for c in cols)

engineered_cols = []

sqft_parts = [c for c in ["TotalBsmtSF", "1stFlrSF", "2ndFlrSF"] if c in df.columns]
if len(sqft_parts) >= 2:
    df["TotalSF"] = df[sqft_parts].sum(axis=1)
    engineered_cols.append("TotalSF")

if safe_cols_exist(["YrSold", "YearBuilt"]):
    df["HouseAge"] = (df["YrSold"] - df["YearBuilt"]).clip(lower=0)
    engineered_cols.append("HouseAge")

if safe_cols_exist(["YrSold", "YearRemodAdd"]):
    df["YearsSinceRemodel"] = (df["YrSold"] - df["YearRemodAdd"]).clip(lower=0)
    engineered_cols.append("YearsSinceRemodel")

if safe_cols_exist(["YearBuilt", "YearRemodAdd"]):
    df["WasRemodeled"] = (df["YearRemodAdd"] != df["YearBuilt"]).astype(int)
    engineered_cols.append("WasRemodeled")

if safe_cols_exist(["OverallQual", "OverallCond"]):
    df["QualCondInteraction"] = df["OverallQual"] * df["OverallCond"]
    engineered_cols.append("QualCondInteraction")

bath_parts = [c for c in ["FullBath", "HalfBath", "BsmtFullBath", "BsmtHalfBath"] if c in df.columns]
if len(bath_parts) >= 2:
    df["TotalBathrooms"] = (
        df.get("FullBath", 0) + df.get("BsmtFullBath", 0)
        + 0.5 * (df.get("HalfBath", 0) + df.get("BsmtHalfBath", 0))
    )
    engineered_cols.append("TotalBathrooms")

if "OverallQual" in df.columns and "TotalSF" in df.columns:
    df["QualSFInteraction"] = df["OverallQual"] * df["TotalSF"]
    engineered_cols.append("QualSFInteraction")

if "PoolArea" in df.columns:
    df["HasPool"] = (df["PoolArea"] > 0).astype(int)
    engineered_cols.append("HasPool")
if "GarageArea" in df.columns:
    df["HasGarage"] = (df["GarageArea"] > 0).astype(int)
    engineered_cols.append("HasGarage")
if "TotalBsmtSF" in df.columns:
    df["HasBasement"] = (df["TotalBsmtSF"] > 0).astype(int)
    engineered_cols.append("HasBasement")

print(f"\nEngineered {len(engineered_cols)} new features: {engineered_cols}")

# ------------------------------------------------------------------
# 4. Split X / y, identify column types, log-transform target
# ------------------------------------------------------------------
df = df.dropna(subset=[TARGET])
y_raw = df[TARGET].astype(float)
X = df.drop(columns=[TARGET])

for id_col in ["Id", "id", "ID"]:
    if id_col in X.columns:
        X = X.drop(columns=[id_col])

num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

print(f"\nTotal columns (excl. target): {X.shape[1]}")
print(f"Numerical columns : {len(num_cols)}")
print(f"Categorical columns: {len(cat_cols)}")

y_log = np.log1p(y_raw)

binary_flag_cols = [c for c in num_cols if X[c].dropna().isin([0, 1]).all()]
skew_candidates = [c for c in num_cols if c not in binary_flag_cols]

# ------------------------------------------------------------------
# 5. Train/test split
# ------------------------------------------------------------------
X_train, X_test, y_train_log, y_test_log, y_train_raw, y_test_raw = train_test_split(
    X, y_log, y_raw, test_size=0.2, random_state=RANDOM_STATE
)
X_train = X_train.reset_index(drop=True)
X_test = X_test.reset_index(drop=True)
y_train_log = y_train_log.reset_index(drop=True)
y_test_log = y_test_log.reset_index(drop=True)
y_train_raw = y_train_raw.reset_index(drop=True)
y_test_raw = y_test_raw.reset_index(drop=True)

# ------------------------------------------------------------------
# 6. Generic fallback imputation (fit on train only)
# ------------------------------------------------------------------
for col in num_cols:
    if X_train[col].isnull().any():
        median_val = X_train[col].median()
        X_train[col] = X_train[col].fillna(median_val)
        X_test[col] = X_test[col].fillna(median_val)

for col in cat_cols:
    if X_train[col].isnull().any():
        mode_val = X_train[col].mode(dropna=True)
        mode_val = mode_val.iloc[0] if len(mode_val) else "Missing"
        X_train[col] = X_train[col].fillna(mode_val)
        X_test[col] = X_test[col].fillna(mode_val)

# ------------------------------------------------------------------
# 7. Log-transform skewed numeric features (train-only skew decision)
# ------------------------------------------------------------------
skew_vals = X_train[skew_candidates].skew()
skewed_num_cols = skew_vals[skew_vals.abs() > 0.75].index.tolist()
skewed_num_cols = [c for c in skewed_num_cols if (X_train[c].min() >= 0)]

print(f"\nLog-transforming {len(skewed_num_cols)} skewed numeric feature(s): {skewed_num_cols}")

for col in skewed_num_cols:
    X_train[col] = np.log1p(X_train[col])
    X_test[col] = np.log1p(X_test[col].clip(lower=0))

# ==================================================================
# FROM HERE: PCA (numeric) + Chi2 (categorical) instead of direct encoding
# ==================================================================
cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
proxy_model = RandomForestRegressor(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1)

# ------------------------------------------------------------------
# 8. Numeric branch — scale, then PCA with optimal n_components
# ------------------------------------------------------------------
scaler = StandardScaler()
X_train_num_scaled = scaler.fit_transform(X_train[num_cols])
X_test_num_scaled = scaler.transform(X_test[num_cols])

candidate_n = list(range(2, min(len(num_cols), 40) + 1, 2))
pca_scores = []
for n in candidate_n:
    pca_tmp = PCA(n_components=n, random_state=RANDOM_STATE)
    X_pca_tmp = pca_tmp.fit_transform(X_train_num_scaled)
    score = cross_val_score(proxy_model, X_pca_tmp, y_train_log, cv=cv, scoring="r2").mean()
    pca_scores.append(score)

best_n_components = candidate_n[int(np.argmax(pca_scores))]
print(f"\n>>> Optimal PCA n_components = {best_n_components}")

final_pca = PCA(n_components=best_n_components, random_state=RANDOM_STATE)
X_train_pca = final_pca.fit_transform(X_train_num_scaled)
X_test_pca = final_pca.transform(X_test_num_scaled)

pca_cols = [f"PC{i+1}" for i in range(best_n_components)]
X_train_pca_df = pd.DataFrame(X_train_pca, columns=pca_cols)
X_test_pca_df = pd.DataFrame(X_test_pca, columns=pca_cols)

explained_var = final_pca.explained_variance_ratio_.sum()
print(f"    ({best_n_components} components capture {explained_var:.1%} of numeric variance)")

# ------------------------------------------------------------------
# 9. Categorical branch — one-hot encode, then Chi2 SelectKBest with optimal k
#    Chi2 requires a categorical target -> SalePrice is binned into quantiles
#    ONLY for this ranking step; the actual regression still uses the real
#    continuous (log) target.
# ------------------------------------------------------------------
ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
X_train_cat_ohe = ohe.fit_transform(X_train[cat_cols])
X_test_cat_ohe = ohe.transform(X_test[cat_cols])

ohe_feature_names = ohe.get_feature_names_out(cat_cols)
X_train_cat_ohe_df = pd.DataFrame(X_train_cat_ohe, columns=ohe_feature_names)
X_test_cat_ohe_df = pd.DataFrame(X_test_cat_ohe, columns=ohe_feature_names)

y_train_binned = pd.qcut(y_train_raw, q=5, labels=False, duplicates="drop")

max_k = X_train_cat_ohe_df.shape[1]
candidate_k = list(range(5, max_k + 1, 5))
if max_k not in candidate_k:
    candidate_k.append(max_k)

chi2_scores = []
for k in candidate_k:
    selector_tmp = SelectKBest(chi2, k=k)
    selector_tmp.fit(X_train_cat_ohe_df, y_train_binned)
    cols_tmp = X_train_cat_ohe_df.columns[selector_tmp.get_support()]
    score = cross_val_score(
        proxy_model, X_train_cat_ohe_df[cols_tmp], y_train_log, cv=cv, scoring="r2"
    ).mean()
    chi2_scores.append(score)

best_k = candidate_k[int(np.argmax(chi2_scores))]
print(f">>> Optimal Chi2 k = {best_k}")

final_selector = SelectKBest(chi2, k=best_k)
final_selector.fit(X_train_cat_ohe_df, y_train_binned)
selected_cat_cols = X_train_cat_ohe_df.columns[final_selector.get_support()]

X_train_cat_selected_df = X_train_cat_ohe_df[selected_cat_cols].reset_index(drop=True)
X_test_cat_selected_df = X_test_cat_ohe_df[selected_cat_cols].reset_index(drop=True)

# ------------------------------------------------------------------
# 10. Combine PCA components + selected categorical dummies
# ------------------------------------------------------------------
X_train_final = pd.concat([X_train_pca_df.reset_index(drop=True), X_train_cat_selected_df], axis=1)
X_test_final = pd.concat([X_test_pca_df.reset_index(drop=True), X_test_cat_selected_df], axis=1)

print(f"\nFinal PCA+Chi2 feature matrix shape: {X_train_final.shape}")

# ------------------------------------------------------------------
# 11. Helpers
# ------------------------------------------------------------------
def adjusted_r2(r2, n_samples, n_features):
    if n_samples - n_features - 1 <= 0:
        print(f"  [WARNING] Adjusted R^2 undefined for n_samples={n_samples}, "
              f"n_features={n_features}. Returning NaN.")
        return np.nan
    return 1 - (1 - r2) * (n_samples - 1) / (n_samples - n_features - 1)


def evaluate_model(model, X_tr, X_te, y_tr_log, y_te_raw, n_features):
    model.fit(X_tr, y_tr_log)
    train_pred_raw = np.expm1(model.predict(X_tr))
    test_pred_raw = np.expm1(model.predict(X_te))

    train_r2 = r2_score(np.expm1(y_tr_log), train_pred_raw)
    test_r2 = r2_score(y_te_raw, test_pred_raw)
    n_test = len(y_te_raw)
    test_adj_r2 = adjusted_r2(test_r2, n_test, n_features)
    gap = train_r2 - test_r2

    return {
        "model": model, "train_r2": train_r2, "test_r2": test_r2,
        "test_adj_r2": test_adj_r2, "train_test_gap": gap,
    }


n_features = X_test_final.shape[1]
results = {}

# ------------------------------------------------------------------
# 12. RandomForest — default, then tuned
# ------------------------------------------------------------------
rf_default = RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1)
results["RandomForest (default)"] = evaluate_model(
    rf_default, X_train_final, X_test_final, y_train_log, y_test_raw, n_features
)

rf_param_dist = {
    "n_estimators": [200, 300, 500, 700, 900],
    "max_depth": [None, 5, 8, 12, 16, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", "log2", 0.5, 0.7, 1.0],
}
rf_search = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1),
    param_distributions=rf_param_dist,
    n_iter=30, cv=cv, scoring="r2", random_state=RANDOM_STATE, n_jobs=-1, verbose=0,
)
rf_search.fit(X_train_final, y_train_log)
print(f"\nBest RandomForest params: {rf_search.best_params_}")
results["RandomForest (tuned)"] = evaluate_model(
    rf_search.best_estimator_, X_train_final, X_test_final, y_train_log, y_test_raw, n_features
)

# ------------------------------------------------------------------
# 13. XGBoost — default, then tuned
# ------------------------------------------------------------------
if HAS_XGB:
    xgb_default = XGBRegressor(random_state=RANDOM_STATE, n_jobs=-1)
    results["XGBoost (default)"] = evaluate_model(
        xgb_default, X_train_final, X_test_final, y_train_log, y_test_raw, n_features
    )

    xgb_param_dist = {
        "n_estimators": [200, 300, 500, 700, 900],
        "max_depth": [3, 4, 5, 6, 8],
        "learning_rate": [0.01, 0.03, 0.05, 0.08, 0.1],
        "subsample": [0.6, 0.7, 0.8, 0.9, 1.0],
        "colsample_bytree": [0.6, 0.7, 0.8, 0.9, 1.0],
        "reg_alpha": [0, 0.01, 0.1, 1],
        "reg_lambda": [0.5, 1, 1.5, 2],
    }
    xgb_search = RandomizedSearchCV(
        estimator=XGBRegressor(random_state=RANDOM_STATE, n_jobs=-1),
        param_distributions=xgb_param_dist,
        n_iter=30, cv=cv, scoring="r2", random_state=RANDOM_STATE, n_jobs=-1, verbose=0,
    )
    xgb_search.fit(X_train_final, y_train_log)
    print(f"Best XGBoost params: {xgb_search.best_params_}")
    results["XGBoost (tuned)"] = evaluate_model(
        xgb_search.best_estimator_, X_train_final, X_test_final, y_train_log, y_test_raw, n_features
    )

# ------------------------------------------------------------------
# 14. Comparison table
# ------------------------------------------------------------------
rows = []
for name, res in results.items():
    rows.append({
        "Model": name,
        "Train R2": round(res["train_r2"], 4),
        "Test R2": round(res["test_r2"], 4),
        "Adjusted R2": round(res["test_adj_r2"], 4) if not np.isnan(res["test_adj_r2"]) else np.nan,
        "Train-Test Gap": round(res["train_test_gap"], 4),
    })
results_df = pd.DataFrame(rows)

def tuning_delta(algo_name):
    d_key, t_key = f"{algo_name} (default)", f"{algo_name} (tuned)"
    if d_key in results and t_key in results:
        return round(results[t_key]["test_adj_r2"] - results[d_key]["test_adj_r2"], 4)
    return np.nan

delta_map = {}
for algo in ["RandomForest", "XGBoost"]:
    delta_map[f"{algo} (default)"] = np.nan
    delta_map[f"{algo} (tuned)"] = tuning_delta(algo)
results_df["Adj R2 Gain from Tuning"] = results_df["Model"].map(delta_map)

print("\n" + "=" * 80)
print("MODEL COMPARISON TABLE (PCA numeric + Chi2 categorical)")
print("=" * 80)
print(results_df.to_string(index=False))
print("=" * 80)

best_row = results_df.loc[results_df["Adjusted R2"].idxmax()]
print(f"\n>>> BEST MODEL by Adjusted R^2: {best_row['Model']} (Adjusted R^2 = {best_row['Adjusted R2']:.4f})")

# ------------------------------------------------------------------
# 15. Benefit / disadvantage discussion for THIS dataset+model combo
# ------------------------------------------------------------------
print("\n" + "=" * 80)
print("PCA + Chi2 — Benefit / Disadvantage for RandomForest & XGBoost on this data")
print("=" * 80)
print(f"""
BENEFITS observed here:
  - Feature count reduced from a full one-hot encoding (likely 200-300+ columns)
    down to {n_features} columns ({best_n_components} PCA components + {best_k}
    chi2-selected categorical dummies) — faster training, smaller Adjusted R^2
    denominator penalty, and no more n_samples-vs-n_features crashes.
  - PCA captured {explained_var:.1%} of numeric variance in {best_n_components}
    components, so little raw numeric signal was discarded.

DISADVANTAGES / risk for THIS model family specifically:
  - RandomForest and XGBoost are tree-based: they split on individual feature
    thresholds and are naturally robust to high dimensionality and correlated
    features. They do NOT need PCA or chi2 to perform well — unlike linear or
    distance-based models, they don't suffer from multicollinearity.
  - PCA components are linear combinations of original numeric features. Trees
    split on axis-aligned thresholds, so a rotated/mixed PC (e.g.
    0.3*GrLivArea - 0.5*YearBuilt + ...) is often HARDER for a tree to exploit
    than the original interpretable variable — this can cost accuracy, not
    just interpretability.
  - Chi2 filtering used a 5-bin quantized version of SalePrice as a proxy
    target for ranking categorical dummies — this discards information before
    the real model ever sees it, versus letting RF/XGBoost learn the true
    continuous relationship directly.
  - Compare the Adjusted R^2 in this table against your earlier DIRECT-ENCODING
    (no PCA/chi2) result on the same data: if PCA+chi2 scores lower, that is
    the expected outcome for tree models on tabular data like this, and
    confirms dimensionality reduction was not needed here.
""")

# ------------------------------------------------------------------
# 16. Feature importance plots — tuned RF vs tuned XGBoost, shown LAST
# ------------------------------------------------------------------
TOP_N = 15

rf_tuned_model = results["RandomForest (tuned)"]["model"]
rf_importances = pd.Series(rf_tuned_model.feature_importances_, index=X_train_final.columns)
rf_top = rf_importances.sort_values(ascending=False).head(TOP_N)

if HAS_XGB:
    xgb_tuned_model = results["XGBoost (tuned)"]["model"]
    xgb_importances = pd.Series(xgb_tuned_model.feature_importances_, index=X_train_final.columns)
    xgb_top = xgb_importances.sort_values(ascending=False).head(TOP_N)

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    axes[0].barh(rf_top.index[::-1], rf_top.values[::-1], color="steelblue")
    axes[0].set_title(f"RandomForest (tuned) — Top {TOP_N} Feature Importances")
    axes[0].set_xlabel("Importance")
    axes[1].barh(xgb_top.index[::-1], xgb_top.values[::-1], color="darkorange")
    axes[1].set_title(f"XGBoost (tuned) — Top {TOP_N} Feature Importances")
    axes[1].set_xlabel("Importance")
    plt.tight_layout()
    plt.savefig("feature_importance_pca_chi2.png", dpi=120)
else:
    plt.figure(figsize=(7, 6))
    plt.barh(rf_top.index[::-1], rf_top.values[::-1], color="steelblue")
    plt.title(f"RandomForest (tuned) — Top {TOP_N} Feature Importances")
    plt.xlabel("Importance")
    plt.tight_layout()
    plt.savefig("feature_importance_pca_chi2.png", dpi=120)

print("\nTop 15 features (RandomForest tuned) — note these are PCs, not raw columns:")
print(rf_top)
if HAS_XGB:
    print("\nTop 15 features (XGBoost tuned):")
    print(xgb_top)

# ------------------------------------------------------------------
# 17. Show all figures at once, at the very end
# ------------------------------------------------------------------
plt.show()
```
